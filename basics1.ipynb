```python
def binary(lst,left,right,key):
    mid=int((right-left)/2)
    if lst[mid]==key:
        return mid+1
    elif lst[mid]>key:
        return binary(lst,mid,right,key)
    else:
        return binary(lst,left,mid,key)
lst=[1,3,4,4,5,5,6]
key=6
res=binary(lst,0,7-1,key)
print(res)
```
