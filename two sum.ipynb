```python
# dictionary
def two_sum(nums, target):
    num_dict = {}
    pairs = []
    
    for i, num in enumerate(nums): #enumerate is used to get a index and the element
        complement = target - num
        if complement in num_dict:
            pairs.append((num_dict[complement], i))
        num_dict[num] = i
        print("==",num_dict)
    print("final",num_dict)
    return pairs

nums = [2, 7, 11, 2, 15, 7]
target = 9
print(two_sum(nums, target))
```
