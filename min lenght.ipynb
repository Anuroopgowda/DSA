```python
lst1=[1,1,2,1] 
lst2=[2,1,1,1]
flag=False
while flag==False:
    left=0
    right=len(lst1)-1
    if lst1==lst2:
        flag=True
        print(len(lst1))
        break
    while lst1[left]==lst2[left]:
        left+=1
    while lst1[right]==lst2[right]:
        right-=1
    lst1=lst1[left:right+1]
    lst2=lst2[left:right+1]
    lst1.sort()
    lst2.sort()
```
