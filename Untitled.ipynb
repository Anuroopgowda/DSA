```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

class LinkedList:
    def __init__(self):
        self.head=None
    
    def insert(self,data):
        newnode=Node(data)
        if self.head==None:
            self.head=newnode
        else:
            newnode.next=self.head
            self.head=newnode
    def display(self):
        cur=self.head
        while cur!=None:
            print(cur.data)
            cur=cur.next
    def delete(self,num):
        count=0
        cur=self.head
        prev=None
        after=cur.next
        while count!=num:
            prev=cur
            cur=cur.next
            after=cur.next
            count+=1
        prev.next=after
        cur.next=None
        
        
            
        
            

l1=LinkedList()
lst=[3,2,1]
for i in lst:
    l1.insert(i)
l1.display()
l1.delete(0)
print("==========")
l1.display()
```
