```python
# we need to reach the end 
# take a goal
# move backwards if the size of jump is greater than goal then change the goal

nums = [3,2,1,0,4]
goal=len(nums)-1
for i in range(len(nums)-1,-1,-1):
    if nums[i]+i>=goal:
        goal=i
        print(goal)
if goal==0:
    print("true")
else:
    print("false")
```
