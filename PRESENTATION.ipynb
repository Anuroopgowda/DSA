Example for RE:

```python
import re

str1 = "My name is Anuroop"
pattern = "op$"  
if re.search(pattern, str1):
    print("Match found: The string ends with 'op'")
else:
    print("No match: The string does not end with 'op'")
```

Slide3
match

```python
pattern = 'hello'
text = 'hello world'

result = re.match(pattern, text)
if result:
    print("Matched:", result.group())
else:
    print("No match")
```

search

```python
pattern = 'world'
text = 'hello world'

result = re.search(pattern, text)
if result:
    print("Found:", result.group())
else:
    print("Not found")
```

findall

```python
pattern = 'My'
text = 'My name is Anuroop.My age is 22'

result = re.findall(pattern, text)
print("All matches:", result)
```

split

```python
pattern = 'is'
text = 'My name is Anuroop.My age is 22'

result = re.split(pattern, text)
print("split result:", result)
```

sub

```python
pattern = 'apples'
replacement = 'oranges'
text = 'I have apples and apples.'

result = re.sub(pattern, replacement, text)
print("After substitution:", result)
```

```python
txt = "hello planet"

x = re.findall("he..o", txt)
print(x)
```

```python
txt = "hello planet"

x = re.findall("^hello", txt)
if x:
  print("Yes, the string starts with 'hello'")
else:
  print("No match")
```

```python
txt = "hello planet"

x = re.findall("hello$", txt)
if x:
  print("Yes, the string ending with 'hello'")
else:
  print("No match")
```

```python
txt = "hello planet"

x = re.findall("he.*o", txt)

print(x)
```

```python
txt = "helo planet"

x = re.findall("he.+o", txt)

print(x)
```

```python
txt = "hello planet"

x = re.findall("he.{2}o", txt)

print(x)
```

```python
txt = "The rain in Spain falls mainly in the plain! stays"

#Check if the string contains either "falls" or "stays":

x = re.findall("falls|stays", txt)

print(x)

if x:
  print("Yes, there is at least one match!")
else:
  print("No match")
```

```python
import re

def validate_pan(pan):
    pattern = '^[A-Z]{5}[0-9]{4}[A-Z]$'
    if re.match(pattern, pan):
        return "Valid PAN Number"
    else:
        return "Invalid PAN Number"

# Example usage
pan_number = input("Enter PAN Number: ")
print(validate_pan(pan_number))
```
