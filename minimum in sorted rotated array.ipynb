```python
#similar to the first one but find the target index
def binarysearch(nums,left,right,target):
    if left>right:
        return -1
    mid=(left+right)//2
    if nums[mid]==target:
        return mid
    elif nums[left]<nums[mid] and nums[left]<=target<=nums[mid-1]:
        return binarysearch(nums,left,mid-1,target)
    elif nums[mid]<nums[right] and nums[mid-1]<=target<=nums[right]:
        return binarysearch(nums,left,mid-1,target)
    elif nums[left]>nums[mid]:
        return binarysearch(nums,left,mid-1,target)
    else:
        return binarysearch(nums,mid+1,right,target)
    
nums = [0,1]
target = 0
result=binarysearch(nums,0,len(nums)-1,target)
print(result)
```
