```python
height = [1,1]
length=len(height)-1
left=0
right=len(height)-1
maxi=0
while left<right:
    height1=min(height[left],height[right])
    area=height1*length
    maxi=max(maxi,area)
    if height[left]<height[right]:
        left+=1
    else:
        right-=1
    length-=1
print(maxi)
    
```
