```python
class Node:
    def __init__(self,data=0,next=None):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
    def insert(self,data):
        newnode=Node(data)
        if self.head is None:
            self.head=newnode
        else:
            cur=self.head
            while cur.next is not None:
                cur=cur.next
            cur.next=newnode
    def display(self):
        cur=self.head
        while cur is not None:
            print(cur.data)
            cur=cur.next
    def sorting(self,l1,l2,l3):
        l3=LinkedList()# new linkedlist
        while l1 and l2:# here l1 and l2 are head of l1 and l2 linkedlist
            if l1.data >l2.data:
                l3.insert(l2.data)
                l2=l2.next
            else:
                l3.insert(l1.data)#call insert function
                l1=l1.next
        while l1:# if remaining
            l3.insert(l1.data)
            l1 = l1.next
        while l2:
            l3.insert(l2.data)
            l2 = l2.next
        return l3
            
        


l1=LinkedList()
lst1=[1,2,3]
for i in lst1:
    l1.insert(i)

l2=LinkedList()
lst2=[4,5,6]
for i in lst2:
    l2.insert(i)



l3=l1.sorting(l1.head,l2.head,l3.head)# return the linkedlist and pass the head value not the whole linkedlist
l3.display()
```
