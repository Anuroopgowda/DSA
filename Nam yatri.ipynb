```python
from collections import deque
n=5
memberId=[6,1,3,8,2]
m=5
friends=[[6,8],[6,1],[1,2],[2,3],[3,8]]
teacher=2
visited={}
graph={}
for mid in memberId:
    graph[mid]=[]
    visited[mid]=-1
print(graph)
# enter friends into the graph
for friend in friends:
    graph[friend[0]].append(friend[1])
    graph[friend[1]].append(friend[0])
print(graph)
print(visited)

queue=deque()
queue.append(teacher)
print(queue)
visited[teacher]=0
while queue:
    expert=queue.popleft()
    curtime=visited[expert]
    for friend in graph[expert]:
        if visited[friend]==-1:
            queue.append(friend)
            print(queue)
            visited[friend]=curtime+1
print(visited)          
print(max(visited.values()))



```
